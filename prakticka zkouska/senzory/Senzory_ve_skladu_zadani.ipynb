# Praktický test – Senzory ve skladu

**Neoficiální cvičné zadání · ILP v Gurobi · obtížnost přibližně jako Knights Placement**

Sklad je rozdělený na obdélníkovou mřížku s **H řádky a W sloupci**. Na některých políčkách stojí stroje, jejichž pozice jsou pevně dané.

Tvým úkolem je pomocí **ILP modelu v Gurobi umístit co nejvíce senzorů**.

Notebook obsahuje zadání a prostor pro tvoje řešení. Neobsahuje hotový model ani nápovědu k jeho formulaci.

## Pravidla

1. Na každém políčku může být nejvýše jeden senzor.
2. Senzor nesmí být na políčku se strojem.
3. Stroje způsobují vibrace: senzor nesmí být ani na políčku **bezprostředně nad, pod, vlevo nebo vpravo od stroje**. Diagonálně vedle stroje být smí.
4. Dva senzory se navzájem ruší, pokud jejich políčka sousedí **vodorovně, svisle nebo diagonálně**. Taková dvojice senzorů je zakázaná.
5. **Maximalizuj celkový počet senzorů.**

Mřížka se na okrajích nepropojuje. Stroje vzájemně sousedit smějí a jejich zakázané oblasti se mohou překrývat.

## Vstup

První řádek obsahuje:

```text
H W B
```

- `H`, `W`: rozměry mřížky, `1 ≤ H, W ≤ 20`.
- `B`: počet strojů, `0 ≤ B ≤ H · W`.

Následuje `B` řádků, každý se souřadnicemi jednoho stroje:

```text
r c
```

**Souřadnice jsou vždy `(řádek, sloupec)` a indexují se od nuly.** Pozice strojů jsou platné a neopakují se.

## Výstup

Na první řádek vypiš maximální počet senzorů `K`. Na dalších `K` řádků vypiš jejich souřadnice:

```text
K
r_1 c_1
r_2 c_2
...
```

Pokud existuje více optimálních rozmístění, můžeš vypsat libovolné. Pořadí souřadnic nehraje roli. Pokud nelze umístit žádný senzor, vypiš pouze `0`.

## Příklad 1

Vstup:

```text
3 4 1
1 1
```

Jeden správný výstup:

```text
4
0 0
0 2
2 0
2 2
```

## Příklad 2

Vstup:

```text
1 1 1
0 0
```

Výstup:

```text
0
```
