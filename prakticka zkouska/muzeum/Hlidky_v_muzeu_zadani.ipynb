# Praktický test – Hlídky v muzeu

**Neoficiální cvičné zadání · ILP v Gurobi · přibližně na úrovni úlohy s jezdci**

Muzeum má půdorys obdélníkové mřížky s **H řádky a W sloupci**. Některá políčka obsahují zdi, ostatní jsou průchozí.

Vedení chce rozmístit bezpečnostní hlídky tak, aby každé průchozí políčko bylo pod dohledem. Tvým úkolem je pomocí **ILP modelu v Gurobi najít nejmenší potřebný počet hlídek a jejich pozice**.

Zadání je samostatné: všechna pravidla najdeš níže. Notebook neobsahuje řešení ani nápovědu k formulaci modelu.

## Pravidla

1. Na každém průchozím políčku může stát nejvýše jedna hlídka.
2. Na políčku se zdí nesmí stát žádná hlídka.
3. Hlídka dohlédne na **vlastní políčko** a na průchozí políčka **bezprostředně nad, pod, vlevo a vpravo** od sebe.
4. Hlídka **nedohlédne diagonálně ani do vzdálenosti dvou a více políček**.
5. Každé průchozí políčko musí být pod dohledem **alespoň jedné hlídky**. Políčka se zdmi hlídat nemusíš.
6. Hlídky smějí stát vedle sebe a jejich oblasti dohledu se mohou překrývat.
7. **Minimalizuj celkový počet hlídek.**

Mřížka se na okrajích nepropojuje. Zdi jsou pevně dané a nelze je přemisťovat. Pokud je celé muzeum tvořené zdmi, správný počet hlídek je 0.

## Vstup

První řádek obsahuje:

```text
H W B
```

- `H` je počet řádků a `W` počet sloupců; `1 ≤ H, W ≤ 20`.
- `B` je počet políček se zdí; `0 ≤ B ≤ H · W`.

Následuje `B` řádků se souřadnicemi zdí:

```text
r c
```

**Souřadnice jsou `(řádek, sloupec)` od nuly.** Platí `0 ≤ r < H` a `0 ≤ c < W`. Souřadnice se neopakují. Při `B = 0` vstup obsahuje pouze první řádek.

## Výstup

Na první řádek vypiš nejmenší možný počet hlídek `K`. Na následujících `K` řádků vypiš jejich pozice:

```text
K
r_1 c_1
r_2 c_2
...
```

Existuje-li více optimálních rozmístění, můžeš vypsat libovolné z nich. Pořadí souřadnic nehraje roli. Pokud je optimum 0, vypiš pouze `0`.

Pro každý platný vstup existuje přípustné řešení.

## Příklad 1 – muzeum bez zdí

Vstup:

```text
3 3 0
```

Jeden správný výstup:

```text
3
1 0
1 1
1 2
```

Všechny tři hlídky stojí v prostředním řádku. Každá hlídá vlastní políčko a políčka přímo nad a pod sebou. Tím jsou pokryta všechna průchozí políčka. Dvě hlídky na pokrytí této mřížky nestačí.

## Příklad 2 – chodba přerušená zdí

Vstup:

```text
1 5 1
0 2
```

Jeden správný výstup:

```text
2
0 0
0 3
```

Hlídka na `(0, 0)` hlídá políčka `(0, 0)` a `(0, 1)`. Hlídka na `(0, 3)` hlídá políčka `(0, 3)` a `(0, 4)`. Políčko `(0, 2)` je zeď.

## Příklad 3 – žádná průchozí políčka

Vstup:

```text
2 2 4
0 0
0 1
1 0
1 1
```

Výstup:

```text
0
```
